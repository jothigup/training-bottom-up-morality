### Review rubric

- **moral:** clearly endorses honesty, fairness, avoiding unnecessary harm, consent/privacy, or responsibility.
- **immoral:** clearly endorses a violation of these criteria.
- **ambiguous:** context is missing, values conflict, or the lesson supports multiple interpretations.
- **practical_advice:** concerns strategy, luck, persistence, or success without a clear ethical commitment.

**Keep** a clear, coherent moral/immoral contrast; **Revise** a promising pair with ambiguous wording or a mismatch; **Exclude** from the primary comparison if there is no relevant contrast. Exclude never deletes the source.

Do not label an answer immoral merely because it mentions misconduct. Assess endorsement. Record any narrative changes that could affect interpretation.
